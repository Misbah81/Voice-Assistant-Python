```python
import subprocess
import os
import speech_recognition as sr
import pyttsx3
import datetime

def speak(text):
    engine = pyttsx3.init()
    engine.say(text)
    engine.runAndWait()

def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        audio = r.listen(source)
        said = ""
        try:
            said = r.recognize_google(audio)
        except Exception as e:
            print("Exception:" + str(e))
    return said

def note(text, directory):
    date = datetime.datetime.now()
    file_name = str(date).replace(":", "-") + "-note.txt"
    if not os.path.exists(directory):
        os.makedirs(directory)
    file_path = os.path.join(directory, file_name)
    with open(file_path, "w") as f:
        f.write(text)
    subprocess.Popen(["notepad.exe", file_path])

WAKE = "hello"
print("Start")

while True:
    print("Listening")
    text = get_audio().lower()
    print(text)

    if WAKE in text:
        speak("I am ready")

        speak("What would you like me to do?")
        command = get_audio().lower()
        print(f"Command: {command}")

        if "write a note" in command:
            speak("Which directory should I save the note in?")
            directory = get_audio().lower()
            print(f"Directory: {directory}")

            if not os.path.exists(directory):
                speak("The directory does not exist. I will create it.")
                os.makedirs(directory)

            speak("What would you like me to write down?")
            note_text = get_audio()
            print(f"Note text: {note_text}")

            note(note_text, directory)
            speak(f"I've made a note of that in the {directory} directory.")
```

```python
import os
import time
import speech_recognition as sr
import pyttsx3
import webbrowser
import urllib.request
import re
import cv2
import numpy as np
import pyautogui

# Read the template image for the "Skip Ad" button
template6 = cv2.imread('C:\\Users\\Misbha\\Desktop\\updated_python_project\\template6.png', 0)

threshold = 0.7

def speak(text):
    engine = pyttsx3.init()
    engine.say(text)
    engine.runAndWait()

def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening...")
        r.adjust_for_ambient_noise(source)
        audio = r.listen(source)
        said = ""
        try:
            said = r.recognize_google(audio)
            print(f"You said: {said}")
        except sr.UnknownValueError:
            print("Sorry, I did not understand that.")
        except sr.RequestError as e:
            print(f"Could not request results; {e}")
        return said.lower()

def play_youtube_video(query):
    search_keyword = query.replace(" ", "-")
    html = urllib.request.urlopen("https://www.youtube.com/results?search_query=" + search_keyword)
    video_ids = re.findall(r"watch\?v=(\S{11})", html.read().decode())
    if not video_ids:
        print("No video found for the search query.")
        speak("No video found for the search query.")
        return
    url = "https://www.youtube.com/watch?v=" + video_ids[0]
    print(url)
    
    # Open the URL in the default web browser
    webbrowser.open(url, new=1)
    
    # Flag to control the loop
    ad_detected = False
    timeout = time.time() + 15  # Set a timeout of 15 seconds

    # Continuous loop to check for YouTube ad
    while time.time() < timeout:
        im1 = pyautogui.screenshot()
        im1 = np.asarray(im1.convert(mode='L'))

        # Checking for template6
        res = cv2.matchTemplate(im1, template6, cv2.TM_CCOEFF_NORMED)
        loc = np.where(res >= threshold)

        # Checking if template is matched
        if loc[0].size != 0:
            # Clicking on the first match
            pyautogui.click(list(zip(*loc[::-1]))[0])
            ad_detected = True
            print("Ad skipped")
            break  # Exit the loop if ad is detected

    # If no ad is detected
    if not ad_detected:
        print("No ad detected")

WAKE = "hello"
SLEEP = "sleep"
SEARCH_STRS = ["search youtube", "search from youtube", "youtube"]
print("Start")

active = False

while True:
    if not active:
        print("Listening for wake word...")
        text = get_audio()
        if WAKE in text:
            speak("I am ready")
            active = True
    else:
        print("Listening for commands...")
        text = get_audio()
        
        if SLEEP in text:
            speak("Going to sleep")
            active = False
            continue

        for phrase in SEARCH_STRS:
            if phrase in text:
                speak("What would you like to search on YouTube?")
                search_query = get_audio()
                play_youtube_video(search_query)
                speak(f"Searching YouTube for {search_query}")
```

```python
import qrcode
import speech_recognition as sr
import pyttsx3

def speak(text):
    engine = pyttsx3.init()
    engine.say(text)
    engine.runAndWait()
    
def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening...")
        audio = r.listen(source)
        said = ""
        try:
            said = r.recognize_google(audio)
            print(f"You said: {said}")
        except sr.UnknownValueError:
            print("Sorry, I did not understand that.")
        except sr.RequestError as e:
            print(f"Could not request results; {e}")
        return said.lower()

def generate_qr_code(data):
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    qr.add_data(data)
    qr.make(fit=True)
    img = qr.make_image(fill='black', back_color='white')
    img.save(f"{data}_qrcode.png")
    print(f"QR code generated and saved as '{data}_qrcode.png'.")

def listen_for_qr_generation():
    speak("What name would you like to generate a QR code for?")
    name = get_audio()
    if name:
        generate_qr_code(name)
        speak(f"The QR code has been generated for {name} as a PNG file.")
    else:
        speak("I didn't hear the name. Please try again.")

WAKE = "hello"
print("Start")
text = get_audio().lower()

while True:
    print("Listening for wake word...")
    text = get_audio()

    if WAKE in text:
        speak("I am ready. What would you like me to do?")
        command = get_audio().lower()

        if "generate qr code" in command:
            listen_for_qr_generation()  # Call the function to generate QR code
        else:
            speak("Sorry, I can only generate QR codes right now.")

        # End the conversation after the task
        speak("Thank you! Bye bye.")
        break  # Exit the loop after completing the task
```

```python
import subprocess
import os
import time
import speech_recognition as sr
import pyttsx3
import pygetwindow as gw

def speak(text):
    engine = pyttsx3.init()
    engine.say(text)
    engine.runAndWait()

def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening...")
        r.adjust_for_ambient_noise(source)
        audio = r.listen(source)
        said = ""
        try:
            said = r.recognize_google(audio)
            print(f"You said: {said}")
        except sr.UnknownValueError:
            print("Sorry, I did not understand that.")
        except sr.RequestError as e:
            print(f"Could not request results; {e}")
        return said.lower()

def create_folder(folder_name):
    try:
        os.makedirs(folder_name)
        speak(f"Folder '{folder_name}' has been created successfully.")
        print(f"Folder '{folder_name}' has been created successfully.")
    except FileExistsError:
        speak(f"Folder '{folder_name}' already exists.")
        print(f"Folder '{folder_name}' already exists.")
    except Exception as e:
        speak(f"An error occurred while creating the folder: {str(e)}")
        print(f"Error: {str(e)}")

def create_python_file(folder_name, file_name):
    file_path = os.path.join(folder_name, f"{file_name}.py")
    with open(file_path, 'w') as file:
        file.write("# Write your Python code here\n")  # Add a placeholder for user code
    speak(f"Python file '{file_name}.py' has been created successfully.")
    print(f"Python file '{file_name}.py' has been created successfully.")

def run_code(folder_name, file_name):
    file_path = os.path.join(folder_name, f"{file_name}.py")
    process = subprocess.Popen(['python', file_path], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    output, error = process.communicate()
    
    if output:
        speak("Code has been successfully executed.")
        print(output.decode())
    if error:
        speak("There was an error while running the code.")
        print(error.decode())

def close_vscode():
    try:
        vscode_window = gw.getWindowsWithTitle("Visual Studio Code")[0]
        vscode_window.close()
        speak("Visual Studio Code has been successfully closed.")
        print("Visual Studio Code has been successfully closed.")
    except IndexError:
        speak("Visual Studio Code is not open.")
        print("Visual Studio Code is not open.")

WAKE = "hello"
print("Start")
active = False
folder_name = ""  # Variable to hold folder name globally
file_name = ""  # Track file name globally
ready = False  # Flag to check if "I am ready" has been said

while True:
    if not active:
        print("Listening for wake word...")
        text = get_audio()
        if WAKE in text and not ready:  # Check if "I am ready" has not been said
            speak("I am ready")
            active = True
            ready = True  # Set ready to True after acknowledging the wake word
    else:
        print("Listening for commands...")
        text = get_audio()

        if "open vs code" in text:
            speak("What would you like to name the folder?")
            folder_name = get_audio().strip()  # Sanitize input
            if folder_name:  # Ensure folder_name is not empty
                create_folder(folder_name)
                # Open VS Code in the specified folder
                subprocess.Popen([r"C:\Program Files\Microsoft VS Code\Code.exe", folder_name])
                
                # Allow some time for VS Code to open
                time.sleep(10)  # Wait 10 seconds for VS Code to fully open
                
                # Print open windows for debugging
                windows = gw.getAllTitles()
                print("Currently open windows:")
                for window in windows:
                    print(window)
                
                # Attempt to get and activate the VS Code window
                try:
                    vscode_window = gw.getWindowsWithTitle("Visual Studio Code")[0]
                    vscode_window.activate()  # Bring VS Code to the foreground
                except IndexError:
                    speak("Could not find Visual Studio Code window to focus.")
                    print("Could not find Visual Studio Code window to focus.")
                
                speak("What would you like me to do next?")
                while True:
                    action = get_audio()

                    if "run the code" in action:
                        speak("Please tell me the name of the Python file to run.")
                        file_name = get_audio().strip()  # Get the file name to run
                        if file_name:  # Ensure file_name is not empty
                            run_code(folder_name, file_name)
                            speak("What would you like me to do next?")
                        else:
                            speak("Please provide a valid file name.")
                    
                    elif "close vs code" in action:
                        close_vscode()
                        break
```

```python
import speech_recognition as sr
from g4f.client import Client
from gtts import gTTS
import playsound
import os
# speech to speech

client = Client()

def get_speech_input():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("Please say something:")
        audio = recognizer.listen(source)
        try:
            text = recognizer.recognize_google(audio)
            print(f"You said: {text}")
            return text
        except sr.UnknownValueError:
            print("Sorry, I could not understand the audio.")
            return None
        except sr.RequestError as e:
            print(f"Could not request results; {e}")
            return None

def get_completion(prompt, client_instance, model="gpt-4o-mini"):
    try:
        messages = [{"role": "user", "content": prompt}]
        response = client_instance.chat.completions.create(
            model=model,
            messages=messages,
            max_tokens=50,
            temperature=0.7,
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"An error occurred: {e}"

def text_to_speech(text, lang='en'):
    tts = gTTS(text=text, lang=lang)
    filename = "response.mp3"
    tts.save(filename)
    playsound.playsound(filename)
    os.remove(filename)

# Get speech input from the user
speech_input = get_speech_input()

# If speech was successfully converted to text, get AI response and convert to speech
if speech_input:
    result = get_completion(speech_input, client)
    print(result)
    text_to_speech(result)
```

```python
import speech_recognition as sr
from g4f.client import Client  # Ensure this is the correct library for your use case
import os
# speech to text
client = Client()

def get_speech_input():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("Please say something:")
        audio = recognizer.listen(source)
        try:
            text = recognizer.recognize_google(audio)
            print(f"You said: {text}")
            return text
        except sr.UnknownValueError:
            print("Sorry, I could not understand the audio.")
            return None
        except sr.RequestError as e:
            print(f"Could not request results; {e}")
            return None

def get_completion(prompt, client_instance, model="gpt-4o-mini"):
    try:
        messages = [{"role": "user", "content": prompt}]
        response = client_instance.chat.completions.create(
            model=model,
            messages=messages,
            max_tokens=250,
            temperature=0.7,
        )
        print("API Response:", response)  # Print the full response for debugging
        if response.choices:
            return response.choices[0].message.content
        else:
            return "No choices returned in the response."
    except Exception as e:
        return f"An error occurred: {e}"

# Get speech input from the user
speech_input = get_speech_input()

# If speech was successfully converted to text, get AI response
if speech_input:
    result = get_completion(speech_input, client)
    print(f"AI Response: {result}")
else:
    print("No valid speech input received.")
```

```python
import qrcode

# Data to encode (my name in this case)
data = "Misbah"

# Create QR code instance
qr = qrcode.QRCode(
    version=1,  # Controls the size of the QR code, 1 is the smallest version
    error_correction=qrcode.constants.ERROR_CORRECT_L,  # Error correction level
    box_size=10,  # Size of each box in the QR code
    border=4,  # Border thickness
)

# Add data to the QR code
qr.add_data(data)
qr.make(fit=True)

# Create the image of the QR code
img = qr.make_image(fill='black', back_color='white')

# Save the image
img.save("misbah_qrcode.png")

print("QR code generated and saved as 'misbah_qrcode.png'")
```

```python
from chatterbot import ChatBot
from chatterbot.trainers import ChatterBotCorpusTrainer
import spacy

# Load spaCy's English NLP model
nlp = spacy.load("en_core_web_sm")

# Create ChatterBot instance
chatbot = ChatBot("SimpleChatBot")

# Train the chatbot on the English corpus
trainer = ChatterBotCorpusTrainer(chatbot)
trainer.train("chatterbot.corpus.english")

def preprocess_input(user_input):
    # NLP Processing to extract entities
    doc = nlp(user_input)
    entities = [(ent.text, ent.label_) for ent in doc.ents]
    return entities

def get_chatbot_response(user_input):
    # Preprocess user input
    entities = preprocess_input(user_input)  # Extract entities
    response = chatbot.get_response(user_input)  # Get response from ChatterBot
    return response, entities

# Simple command-line interface for the chatbot
if __name__ == "__main__":
    print("Welcome to the Simple ChatBot! Type 'exit' to stop.")
    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            print("ChatBot: Goodbye!")
            break
        
        response, entities = get_chatbot_response(user_input)
        print(f"ChatBot: {response} (Entities detected: {entities})")
```
